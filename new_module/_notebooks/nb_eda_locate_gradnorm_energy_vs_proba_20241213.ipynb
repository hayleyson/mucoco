```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    average_precision_score,
    f1_score,
    fbeta_score,
    precision_score,
    recall_score,
)
```

```python
def apply_recall(row, binary_labels_col:str, binary_preds_col:str):
    return recall_score(row[binary_labels_col],row[binary_preds_col], zero_division=np.nan)
```

```python
labels = pd.read_json('/data/hyeryung/mucoco/new_module/data/NLI_locate/nli_contra_300_locate_labels.jsonl', lines=True)
labels = labels[['pairID', 'hypothesis_token_labels_binary']].copy()
use_energy_result = pd.read_json('/data/hyeryung/loc_edit/models/nli/roberta_large_snli_mnli_anli_train_dev_with_finegrained_finegrained_labels_cross_entropy_n_a/zgs9e2sr/nli_contra_300_locate_result_pearsonr_energy.jsonl', lines=True)
use_proba_result = pd.read_json('/data/hyeryung/loc_edit/models/nli/roberta_large_snli_mnli_anli_train_dev_with_finegrained_finegrained_labels_cross_entropy_n_a/zgs9e2sr/nli_contra_300_locate_result_pearsonr_proba.jsonl', lines=True)
# use_proba_result = pd.read_json('/data/hyeryung/loc_edit/models/nli/roberta_large_snli_mnli_anli_train_dev_with_finegrained_binary_labels_binary_cross_entropy_n_a/1731247443//nli_contra_300_locate_result_pearsonr_proba.jsonl', lines=True)
use_energy_result = pd.merge(labels, use_energy_result, on='pairID')
use_energy_result['recall'] = use_energy_result.apply(lambda x: apply_recall(x, 'hypothesis_token_labels_binary', 'hypothesis_token_pred_binary'), axis=1)
use_proba_result = pd.merge(labels, use_proba_result, on='pairID')
use_proba_result['recall'] = use_proba_result.apply(lambda x: apply_recall(x, 'hypothesis_token_labels_binary', 'hypothesis_token_pred_binary'), axis=1)
use_energy_result['num_mask'] = use_energy_result['hypothesis_token_pred_binary'].apply(sum)
use_proba_result['num_mask'] = use_proba_result['hypothesis_token_pred_binary'].apply(sum)
merged_result = pd.merge(use_proba_result[['pairID', 'recall', 'num_mask']], use_energy_result[['pairID', 'recall', 'num_mask']], on='pairID', suffixes=('_proba', '_energy'))
display(merged_result.loc[merged_result['recall_proba'] > merged_result['recall_energy'], ['recall_proba', 'recall_energy', 'num_mask_proba', 'num_mask_energy']])
display(merged_result.loc[merged_result['num_mask_proba'] > merged_result['num_mask_energy'],:])
display(merged_result.loc[merged_result['num_mask_proba'] < merged_result['num_mask_energy'],:])
display(merged_result.loc[merged_result['num_mask_proba'] == merged_result['num_mask_energy'],:])
use_energy_result['num_mask'].quantile([x/10 for x in range(11)])
use_proba_result['num_mask'].quantile([x/10 for x in range(11)])
```
