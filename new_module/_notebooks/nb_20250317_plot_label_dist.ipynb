```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data1=pd.read_json('data/toxicity/jigsaw-unintended-bias-in-toxicity-classification/fine-grained/train.jsonl',lines=True)
data2=pd.read_json('data/nli/snli_mnli_anli_train_dev_with_finegrained.jsonl',lines=True)
data3=pd.read_json('data/yelp/train.jsonl',lines=True)
```

```python
data3['labels']=data3['labels'].apply(lambda x: (x-1)/4)
```

```python
# configure matplotlib such that it embeds TrueType fonts when saving figures as PDF files.
plt.rcParams['pdf.fonttype'] = 42

# Create a 3-panel figure
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Panel 1: Distribution of all labels
axes[0].hist(data1["labels"], bins=np.arange(0, 1.1, 0.05), color="skyblue", edgecolor="black")
# axes[0].set_title("Toxicity Avoidance", fontsize=18)
axes[0].set_title("Toxicity", fontsize=18)
axes[0].set_xlabel("Label Value",  fontsize=18)
axes[0].set_ylabel("Count",  fontsize=18)

# Panel 2: Distribution of binary labels
axes[1].hist(data2["finegrained_labels"], bins=np.arange(0, 1.1, 0.05), color="salmon", edgecolor="black")
# axes[1].set_title("Logical Inconsistency Avoidance", fontsize=18)
axes[1].set_title("Logical Inconsistency", fontsize=18)
axes[1].set_xlabel("Label Value", fontsize=18)

# Panel 3: Distribution of regression labels
axes[2].hist(data3["labels"], bins=np.arange(0, 1.1, 0.05), color="lightgreen", edgecolor="black")
# axes[2].set_title("Sentiment Control", fontsize=18)
axes[2].set_title("Positivity", fontsize=18)
axes[2].set_xlabel("Label Value", fontsize=18)

# Adjust layout and show plot
plt.tight_layout()
# plt.show()
plt.savefig("new_module/_notebooks/outputs/em_label_dist/em_training_label_dist_enlarged.pdf", dpi=300, bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator

# Create a 3-panel figure vertically
fig, axes = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

# Data placeholders
# data1, data2, data3 assumed to be defined

# Shared settings for y-axis
for ax in axes:
    ax.yaxis.set_major_locator(MaxNLocator(nbins=4, integer=True))  # 3~4 y-ticks
    ax.set_ylabel("Count", labelpad=15, rotation=90, ha='right', va='center', fontsize=14)
    ax.tick_params(axis='both', labelsize=12)
# Panel 1
axes[0].hist(data1["labels"], bins=np.arange(0, 1.1, 0.05), color="skyblue", edgecolor="black")

# Panel 2
axes[1].hist(data2["finegrained_labels"], bins=np.arange(0, 1.1, 0.05), color="salmon", edgecolor="black")

# Panel 3
axes[2].hist(data3["labels"], bins=np.arange(0, 1.1, 0.05), color="lightgreen", edgecolor="black")
axes[2].set_xlabel("Label Value", fontsize=14)

# Reduce vertical spacing between subplots
fig.align_ylabels(axes)
plt.subplots_adjust(left=0.1, hspace=0.1)
# plt.show()

# plt.show()
# plt.savefig("new_module/_notebooks/outputs/em_label_dist/em_training_label_dist_stacked.png", dpi=300, bbox_inches='tight')
# plt.close()
```
