```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from new_module.dev_utils.utils import *
```

```python
qwen_result = read_outputs("outputs/nli/1vzneid4/outputs_epsilon0.99.txt")
```

```python
qwen_result['e_f'] = qwen_result['losses'].apply(lambda x: x[0])
qwen_result['e_c'] = qwen_result['losses'].apply(lambda x: x[1])
```

```python
fig = plt.figure()
ax = plt.gca()
sns.kdeplot(qwen_result, x='e_f', ax=ax, label='e_f')
sns.kdeplot(qwen_result, x='e_c', ax=ax, label='e_c')
ax.legend()
plt.show()
```

```python
sns.scatterplot(qwen_result, x='e_f', y='e_c')
```
