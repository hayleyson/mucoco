```python
import pandas as pd
```

```python
traindev = pd.read_json('/data/hyeryung/mucoco/data/nli/snli_mnli_anli_train_dev_with_finegrained.jsonl', lines=True)
```

```python
def clean_source_value(x):
    if 'anli' in x:
        names = x.split('_')
        return names[0] + '_' + names[-1]
    elif 'mnli' in x:
        return x.split('_')[0]
    else:
        return x
traindev['source2'] = traindev['source'].apply(clean_source_value)   
```

```python
traindev.groupby(['split', 'source2']).size()
```
