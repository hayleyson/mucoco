### COLA drop / LLM Edit with Original / toxicity avoidance task

```python
## cola 파일과 생성문 파일을 읽기 -> cola가 낮은 생성문 확인하기
llmlewith = read_outputs('/data/hyeryung/mucoco/outputs/llmedit/dump/iter_loc_edit_qwen/edited/2_tox_edited_38592.jsonl_total_0')
llmlewo = read_outputs('/data/hyeryung/mucoco/outputs/llmedit/results_2025/tox_iter_1_prompt_ablation/wo_orig/fluency/2_tox_loc_edit_50269.jsonl')
masked_llmlewith = read_outputs('/data/hyeryung/mucoco/outputs/llmedit/dump/iter_loc_edit_qwen/located/2_tox_located_38592.jsonl_filtered_0')
cola_llmlewith = read_metric_file('/data/hyeryung/mucoco/outputs/llmedit/dump/iter_loc_edit_qwen/edited_fluency/2_tox_edited_38592.jsonl_total_0-results.txt.fluency', 'fluency')
```

```python
llmlewith['masked_text'] = masked_llmlewith['text']
llmlewith['llmlewo'] = llmlewo['text']
llmlewith['fluency_proba'] = cola_llmlewith['fluency_proba']
llmlewith.sort_values(by='fluency_proba', ascending=True, inplace=True)
```

```python
num_examples = 20
for i, (idx, row) in enumerate(llmlewith.iterrows()):
    if i >= num_examples:
        break
    print(f"index: {idx}")
    print(f"fluency proba: {row['fluency_proba']}")
    print(f"prompt: {row['prompt']}")
    print(f"masked_text: {row['masked_text']}")
    print(f"text: {row['text']}")
    print(f"text by LLM LE w/o Original: {row['llmlewo']}")
    print("")
```
