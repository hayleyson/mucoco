Gemma

```python
json_data = read_metric_file('/data/hyeryung/mucoco/outputs/llmedit/dump/iter_loc_edit_llm/edited_rmv_em/6_nli_edited_37635.jsonl_total_0-results.txt.nli', 'nli')
```

```python
def determine_class(row):
    if row['entailment_prob'] > row['neutral_prob'] and row['entailment_prob'] > row['contradiction_prob']:
        return 'entailment'
    elif row['neutral_prob'] > row['entailment_prob'] and row['neutral_prob'] > row['contradiction_prob']:
        return 'neutral'
    elif row['contradiction_prob'] > row['entailment_prob'] and row['contradiction_prob'] > row['neutral_prob']:
        return 'contradiction'
    else:
        return 'tie exists'
json_data['nli_class'] = json_data.apply(determine_class, axis=1)
```

```python
json_data['nli_class'].value_counts()
```

```python
json_data['nli_class'].apply(lambda x: x == 'contradiction').mean()
```
