```python
import pandas as pd
import os
def ravel(unraveled_df):
    if 'tokens' in unraveled_df:
        unraveled_df['generations']= unraveled_df.apply(lambda x: [{'text': x['text'],
                                                               'tokens': x['tokens']}],axis=1)
    else:
        unraveled_df['generations']= unraveled_df.apply(lambda x: [{'text': x['text']}],axis=1)
    return_df = unraveled_df.groupby('prompt')['generations'].sum([]).reset_index()
    return_df['prompt'] = return_df['prompt'].apply(lambda x: {'text':x})
        
    return return_df
```

### 길이 50 처리

```python

## prompt들 리스트를 읽어와서, 
prompts = """Once upon a time
The book
The chicken
The city
The country
The horse
The lake
The last time
The movie
The painting
The pizza
The potato
The president of the country
The road
The year is 1910""".split("\n")
```

```python

## prompted_gen 폴더 아래에 len_50 으로 있는 문장들을 읽어온다.
aggregated_data = []
for (root, dirs, files) in os.walk("outputs/sentiment/mixmatch/prompted_gen/"):
    # print(root)
    # print(dirs)
    # print(files)
    # print('-'*20)
    if ('opt_samples.txt' in files) and ('len_50' in root):
        with open(os.path.join(root, 'opt_samples.txt'), 'r') as f:
            data = f.readlines()
            aggregated_data.extend(data)
```

```python
print(len(aggregated_data))
```

```python
## 위에서 읽어온 문장의 prompt를 match해서 prompt를 분리해낸다.
matched_prompts = []
generations = []
for line in aggregated_data:
    found_prompt = False
    for prompt in prompts:
        if line.startswith(prompt.lower()):
            matched_prompts.append(prompt)
            generations.append(line[len(prompt):].strip('\n'))
            found_prompt = True
    if not found_prompt:
        print(line)
        matched_prompts.append("")
```

```python
## prompt별로 몇건씩 생성되었는지 체크한다. 
df = pd.DataFrame({'prompt': matched_prompts, 'text': generations})
df.groupby('prompt').size() ## we expect 20 samples per prompt, and all prompts satisfy this
```

```python
## prompt별로 20건씩만 샘플링한다.
df_len50 = df.groupby('prompt').sample(20, random_state=42)
df_len50 = ravel(df_len50)
```
