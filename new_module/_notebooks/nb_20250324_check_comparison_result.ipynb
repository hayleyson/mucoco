### 전체 데이터에서 violation 이 일어나는 건수 확인

```python
import os, json, random, pickle
import pandas as pd
from openai import OpenAI

def unravel(outputs_df):
    outputs_df=outputs_df.explode('generations',ignore_index=True)
    outputs_df['prompt']=outputs_df['prompt'].apply(lambda x: x['text'])
    outputs_df['generations']=outputs_df['generations'].apply(lambda x: x['text'] if isinstance(x, dict) else x)
    outputs_df = outputs_df.dropna().reset_index(drop=True)
    return outputs_df

def read_outputs(file_path):
    return unravel(pd.read_json(file_path, lines=True))


orig_path = 'gpt-3.5-turbo-0125_realtoxicityprompts_0shot_150_below_nontoxic_threshold_0_95_332.jsonl'
# LLM Edit WO Locate
llm_edit_wo_path = 'llm_edit_wo_locate_5_tox_loc_edit_38576.jsonl'
# L&E's
## Mask&Infill
loc_edit_mlm_path = 'loc_edit_mlm_outputs_epsilon0.95.txt.0'
## LLM Edit (버전 아직 안정해졌음)
loc_edit_llm_1_path = 'loc_edit_llm_both_2_tox_edited_38592.jsonl_total_0'
loc_edit_llm_2_path = 'loc_edit_llm_masked_2_tox_loc_edit_50269.jsonl'

original = read_outputs(orig_path)

llm_edit_wo_loc = read_outputs(llm_edit_wo_path)

mask_and_infill = read_outputs(loc_edit_mlm_path)

loc_edit_llm1 = read_outputs(loc_edit_llm_1_path)
loc_edit_llm2 = read_outputs(loc_edit_llm_2_path)
```

```python
compr_df = pd.DataFrame({
    'prompt': original['prompt'],
    'original': original['generations'],
    'llm_edit_wo_loc': llm_edit_wo_loc['generations'],
    'mask_and_infill': mask_and_infill['generations'],
    'loc_edit_llm1': loc_edit_llm1['generations'],
    'loc_edit_llm2': loc_edit_llm2['generations'],
})
```

```python
def check_violation(prompt, generation):
    return (prompt.strip() not in generation) and (tuple(prompt.strip().split()[:3]) == tuple(generation.strip().split()[:3]))
```

```python
compr_df['llm_edit_wo_loc_violation'] = compr_df.apply(lambda x: check_violation(x['prompt'], x['llm_edit_wo_loc']), axis=1)
compr_df['mask_and_infill_violation'] = compr_df.apply(lambda x: check_violation(x['prompt'], x['mask_and_infill']), axis=1)
compr_df['loc_edit_llm1_violation'] = compr_df.apply(lambda x: check_violation(x['prompt'], x['loc_edit_llm1']), axis=1)
compr_df['loc_edit_llm2_violation'] = compr_df.apply(lambda x: check_violation(x['prompt'], x['loc_edit_llm2']), axis=1)
```

### 가설 검증: violation 일 때 content preservation이 더 높나?

Yes. 그런데 (baseline not vio, ours vio) 케이스는 적어서, (baseline vio, ours not vio) 의 반대의 경우에도 성립해야 가장 깔끔한데 반대가 성립한다는 결론을 내리기가 조금 어렵다.

```python
def count_case_by_violation_yn(ours_name, arg_test_cases, arg_eval_indices, arg_llmEdit_vs_ours):

    # var_summary dictionary 생성
    var_summary = {'content_prsv': {'baseline_vio,ours_not_vio': {ours_name: 0, 'Tie': 0, 'llm_edit_wo_locate': 0},
                                'baseline_not_vio,ours_vio': {ours_name: 0, 'Tie': 0, 'llm_edit_wo_locate': 0},
                                'both_not_vio': {ours_name: 0, 'Tie': 0, 'llm_edit_wo_locate': 0},
                                'both_vio': {ours_name: 0, 'Tie': 0, 'llm_edit_wo_locate': 0}}}

    # 테케 중 첫 100개로 결과 내기 (이 첫 100개가 paper에 들어간 비교 대상과 같음)
    for index in range(100):

        # 필요 정보 읽어오기 
        prompt = arg_test_cases[str(arg_eval_indices[index])]['prompt']
        original = arg_test_cases[str(arg_eval_indices[index])]['original_generation']
        llm_edit_wo_loc = arg_test_cases[str(arg_eval_indices[index])]['llm_edit_wo_locate']
        ours = arg_test_cases[str(arg_eval_indices[index])][ours_name+'_generation'] if ours_name == 'mask_and_infill' else arg_test_cases[str(arg_eval_indices[index])][ours_name] 
        
        # violation 인지 여부 체크 
        vio_llm_edit_wo_loc = (prompt.strip() not in llm_edit_wo_loc) and (tuple(prompt.strip().split()[:3]) == tuple(llm_edit_wo_loc.strip().split()[:3]))
        vio_ours = (prompt.strip() not in ours) and (tuple(prompt.strip().split()[:3]) == tuple(ours.strip().split()[:3]))
        
        # 각 case 별로 나눠서 winner 저장
        if (vio_llm_edit_wo_loc) and (not vio_ours):
            winner = arg_llmEdit_vs_ours[str(arg_eval_indices[index])]['content_prsv']['winner']
            var_summary['content_prsv']['baseline_vio,ours_not_vio'][winner] += 1
            
        elif (not vio_llm_edit_wo_loc) and (vio_ours):
            winner = arg_llmEdit_vs_ours[str(arg_eval_indices[index])]['content_prsv']['winner']
            var_summary['content_prsv']['baseline_not_vio,ours_vio'][winner] += 1
            
        elif (vio_llm_edit_wo_loc) and (vio_ours):
            winner = arg_llmEdit_vs_ours[str(arg_eval_indices[index])]['content_prsv']['winner']
            var_summary['content_prsv']['both_vio'][winner] += 1
            
        else:
            winner = arg_llmEdit_vs_ours[str(arg_eval_indices[index])]['content_prsv']['winner']
            var_summary['content_prsv']['both_not_vio'][winner] += 1
    
    
    
    return pd.DataFrame(var_summary['content_prsv']).T[['llm_edit_wo_locate', ours_name, 'Tie']]
```

### 사례 확인: 한쪽, 양쪽 violation 일 때 판단 결과

```python
with open('/data/hyeryung/mucoco/new_module/qualitative_eval/TACL_2025/test_cases_loc_edit_llm.json', 'r') as f:
    test_cases = json.load(f)
    
with open('/data/hyeryung/mucoco/new_module/qualitative_eval/TACL_2025/eval_indices_loc_edit_llm.pickle', 'rb') as f:
    eval_indices = pickle.load(f)

with open('/data/hyeryung/mucoco/new_module/qualitative_eval/TACL_2025/gpt4o_results/(6)w_prefix-llmEdit_vs_loc_edit_llm1.json', 'r') as f:
    llmEdit_vs_ours = json.load(f)
```

```python
def get_index():
    i = 0
    while i < 100:
        yield eval_indices[i]
        i += 1
        
index_generator = get_index()
    
def get_oneside_violation_item_to_inspect(task, ours_name, desired_winner, width=100):

    yield_next = True
    while yield_next:
        try:
            index = next(index_generator)
        except StopIteration:
            print("No more items to inspect.")
            return
        prompt = test_cases[str(index)]['prompt']
        original = test_cases[str(index)]['original_generation']
        llm_edit_wo_loc = test_cases[str(index)]['llm_edit_wo_locate']
        ours = test_cases[str(index)][ours_name+'_generation'] if ours_name == 'mask_and_infill' else test_cases[str(index)][ours_name] 
        # input_for_llm = llmEdit_vs_ours[str(index)][task]['input_for_llm']
        winner = llmEdit_vs_ours[str(index)][task]['winner']
        
        vio_llm_edit_wo_loc = (prompt.strip() not in llm_edit_wo_loc) and (tuple(prompt.strip().split()[:3]) == tuple(llm_edit_wo_loc.strip().split()[:3]))
        vio_ours = (prompt.strip() not in ours) and (tuple(prompt.strip().split()[:3]) == tuple(ours.strip().split()[:3]))

        
        if (vio_llm_edit_wo_loc and vio_ours) or (not vio_llm_edit_wo_loc and not vio_ours) or (winner != desired_winner):
            yield_next = True
            continue
        else:
            yield_next = False
        
        if prompt not in llm_edit_wo_loc:
            if tuple(prompt.strip().split()[:3]) != tuple(llm_edit_wo_loc.strip().split()[:3]):
                llm_edit_wo_loc = prompt + " " + llm_edit_wo_loc
        if prompt not in ours:
            if tuple(prompt.strip().split()[:3]) != tuple(ours.strip().split()[:3]):
                ours = prompt + " " + ours
        
        explanation = llmEdit_vs_ours[str(index)][task]['naive_llm_output']
        print(f'Index: {index}')
        print(f'Prompt: {prompt}', width=width)
        print('-'*20)
        print(f'Original: {original}', width=width)
        print(f'LLM Edit w/o Locate: {llm_edit_wo_loc}', width=width)
        print(f'Ours: {ours}', width=width)
        print('-'*20)
        # print(f'Input for LLM: {input_for_llm}', width=width)
        print(f'Winner: {winner}', width=width)
        print(f'Explanation: {explanation}', width=width)
        

def get_both_not_violation_item_to_inspect(task, ours_name, desired_winner, width=100):

    yield_next = True
    while yield_next:
        try:
            index = next(index_generator)
        except StopIteration:
            print("No more items to inspect.")
            return
        prompt = test_cases[str(index)]['prompt']
        original = test_cases[str(index)]['original_generation']
        llm_edit_wo_loc = test_cases[str(index)]['llm_edit_wo_locate']
        ours = test_cases[str(index)][ours_name+'_generation'] if ours_name == 'mask_and_infill' else test_cases[str(index)][ours_name] 
        # input_for_llm = llmEdit_vs_ours[str(index)][task]['input_for_llm']
        winner = llmEdit_vs_ours[str(index)][task]['winner']
        
        vio_llm_edit_wo_loc = (prompt.strip() not in llm_edit_wo_loc) and (tuple(prompt.strip().split()[:3]) == tuple(llm_edit_wo_loc.strip().split()[:3]))
        vio_ours = (prompt.strip() not in ours) and (tuple(prompt.strip().split()[:3]) == tuple(ours.strip().split()[:3]))

        
        if (vio_llm_edit_wo_loc) or (vio_ours) or (winner != desired_winner):
            yield_next = True
            continue
        else:
            yield_next = False
        
        if prompt not in llm_edit_wo_loc:
            if tuple(prompt.strip().split()[:3]) != tuple(llm_edit_wo_loc.strip().split()[:3]):
                llm_edit_wo_loc = prompt + " " + llm_edit_wo_loc
        if prompt not in ours:
            if tuple(prompt.strip().split()[:3]) != tuple(ours.strip().split()[:3]):
                ours = prompt + " " + ours
        
        explanation = llmEdit_vs_ours[str(index)][task]['naive_llm_output']
        print(f'Index: {index}')
        print(f'Prompt: {prompt}', width=width)
        print('-'*20)
        print(f'Original: {original}', width=width)
        print(f'LLM Edit w/o Locate: {llm_edit_wo_loc}', width=width)
        print(f'Ours: {ours}', width=width)
        print('-'*20)
        # print(f'Input for LLM: {input_for_llm}', width=width)
        print(f'Winner: {winner}', width=width)
        print(f'Explanation: {explanation}', width=width)
```

### LLM Edit wo Locate v.s. L&E LLM Edit w/ Original

```python
class ItemGetter():
    
    def __init__(self, test_cases, eval_indices, llmEdit_vs_ours):
        self.test_cases = test_cases
        self.eval_indices = eval_indices
        self.llmEdit_vs_ours = llmEdit_vs_ours
        self.index_generator = self.get_index()
        
    def get_index(self):
        i = 0
        while i < 100:
            yield self.eval_indices[i]
            i += 1

    def get_nontie_item_to_inspect(self, task, ours_name, width=100):

        yield_next = True
        while yield_next:
            try:
                index = next(self.index_generator)
            except StopIteration:
                print("No more items to inspect.")
                return
            prompt = self.test_cases[str(index)]['prompt']
            original = self.test_cases[str(index)]['original_generation']
            llm_edit_wo_loc = self.test_cases[str(index)]['llm_edit_wo_locate']
            ours = self.test_cases[str(index)][ours_name]
            # input_for_llm = llmEdit_vs_maskInfill[str(index)][task]['input_for_llm']
            winner = self.llmEdit_vs_ours[str(index)][task]['winner']
            if winner == 'Tie':
                yield_next = True
                continue
            else:
                yield_next = False
                
            if prompt not in llm_edit_wo_loc:
                if tuple(prompt.strip().split()[:3]) != tuple(llm_edit_wo_loc.strip().split()[:3]):
                    llm_edit_wo_loc = prompt + " " + llm_edit_wo_loc
            if prompt not in ours:
                if tuple(prompt.strip().split()[:3]) != tuple(ours.strip().split()[:3]):
                    ours = prompt + " " + ours
            explanation = self.llmEdit_vs_ours[str(index)][task]['naive_llm_output']
            print(f'Index: {index}')
            print(f'Prompt: {prompt}', width=width)
            print('-'*20)
            print(f'Original: {original}', width=width)
            print(f'LLM Edit w/o Locate: {llm_edit_wo_loc}', width=width)
            print(f'L&E LLM Edit Both: {ours}', width=width)
            print('-'*20)
            # print(f'Input for LLM: {input_for_llm}', width=width)
            print(f'Winner: {winner}', width=width)
            print(f'Explanation: {explanation}', width=width)
```
