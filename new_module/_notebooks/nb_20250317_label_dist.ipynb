```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data1=pd.read_json('data/toxicity/jigsaw-unintended-bias-in-toxicity-classification/fine-grained/train.jsonl',lines=True)
data2=pd.read_json('data/nli/snli_mnli_anli_train_dev_with_finegrained.jsonl',lines=True)
data3=pd.read_json('data/yelp/train.jsonl',lines=True)
```

```python
data3['labels']=data3['labels'].apply(lambda x: (x-1)/4)
```

```python
# Create a 3-panel figure
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Panel 1: Distribution of all labels
axes[0].hist(data1["labels"], bins=np.arange(0, 1.1, 0.05), color="skyblue", edgecolor="black")
axes[0].set_title("Toxicity Avoidance")
axes[0].set_xlabel("Label Value")
axes[0].set_ylabel("Count")

# Panel 2: Distribution of binary labels
axes[1].hist(data2["finegrained_labels"], bins=np.arange(0, 1.1, 0.05), color="salmon", edgecolor="black")
axes[1].set_title("Logical Inconsistency Avoidance")
axes[1].set_xlabel("Label Value")

# Panel 3: Distribution of regression labels
axes[2].hist(data3["labels"], bins=np.arange(0, 1.1, 0.05), color="lightgreen", edgecolor="black")
axes[2].set_title("Sentiment Control")
axes[2].set_xlabel("Label Value")

# Adjust layout and show plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define bin range
bins = np.arange(0, 1.1, 0.05)

# Panel 1: Toxicity Avoidance
plt.figure(figsize=(5, 5))
plt.hist(data1["labels"], bins=bins, color="skyblue", edgecolor="black")
# plt.title("Toxicity Avoidance")
plt.xlabel("Label Value")
plt.ylabel("Count")
plt.savefig("new_module/_notebooks/outputs/em_label_dist/toxicity_avoidance.png", dpi=300, bbox_inches='tight')
plt.close()

# Panel 2: Logical Inconsistency Avoidance
plt.figure(figsize=(5, 5))
plt.hist(data2["finegrained_labels"], bins=bins, color="salmon", edgecolor="black")
# plt.title("Logical Inconsistency Avoidance")
plt.xlabel("Label Value")
plt.ylabel("Count")
plt.savefig("new_module/_notebooks/outputs/em_label_dist/logical_inconsistency_avoidance.png", dpi=300, bbox_inches='tight')
plt.close()

# Panel 3: Sentiment Control
plt.figure(figsize=(5, 5))
plt.hist(data3["labels"], bins=bins, color="lightgreen", edgecolor="black")
# plt.title("Sentiment Control")
plt.xlabel("Label Value")
plt.ylabel("Count")
plt.savefig("new_module/_notebooks/outputs/em_label_dist/sentiment_control.png", dpi=300, bbox_inches='tight')
plt.close()
```
