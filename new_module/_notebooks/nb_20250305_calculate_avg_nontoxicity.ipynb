```python
import pandas as pd

def unravel_toxicity_data(df):
    df['toxicity']=df['allresponses'].apply(lambda x: [x[0]['attributeScores']['TOXICITY']['summaryScore']['value'] for x in list(x.values())])
    df=df.explode('toxicity',ignore_index=True)
    return df
```

```python
## 파일 읽어오기
### energy model 
em = pd.read_json('outputs/toxicity/llm/j18pi8ab/outputs_epsilon0.95.txt.0-results.txt.toxicity', lines=True)
em = unravel_toxicity_data(em)
### classifier
clsf = pd.read_json('outputs/toxicity/llm/cqlac17z/results_epsilon0.998671-test.txt.toxicity', lines=True)
clsf = unravel_toxicity_data(clsf)

## nontoxicity로 치환
em['nontoxicity'] = 1 - em['toxicity']
clsf['nontoxicity'] = 1 - clsf['toxicity']

## 평균 계산
print(f"regressor avg. non toxicity: {em['nontoxicity'].mean()}")
print(f"classifier avg. non toxicity: {clsf['nontoxicity'].mean()}")
```
