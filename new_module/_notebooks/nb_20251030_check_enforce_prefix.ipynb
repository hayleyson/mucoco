```python
def remove_prompt(row):
    prompt = normalize_quotation(row['prompt'])
    text = normalize_quotation(row['text'])
    if prompt in text:
        return text.replace(prompt, '')
    else:
        return row['text']

def check_prompt(row):
    prompt = normalize_quotation(row['prompt'])
    text = normalize_quotation(row['text'])    
    if prompt in text:
        return 1
    if ' '.join(prompt.split(' ')[:3]) in text:
        return -1
    else:
        return 0
    
def normalize_quotation(text):
    
    text = text.replace("'", '"')
    text = text.replace("’", '"')
    text = text.replace('“', '"')
    return text
    
outputs['continuation'] = outputs.apply(remove_prompt, axis=1)
outputs['prompt_repeated'] = outputs.apply(check_prompt, axis=1)
```
