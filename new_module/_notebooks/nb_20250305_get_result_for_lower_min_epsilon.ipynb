```python
## 0.9 셋은 이미 만들어놓았음
## 0.9 셋은 0.99 셋의 부분집합임
with open('new_module/llm_experiments/generate_with_llm/baselm_gens/gpt-3.5-turbo-0125/nontoxic/gpt-3.5-turbo-0125_realtoxicityprompts_0shot_150_below_nontoxic_threshold_0_95_332_index.txt', 'r') as f:
    old_index = f.read().split()
    old_index = [int(i) for i in old_index]
with open('new_module/llm_experiments/generate_with_llm/baselm_gens/gpt-3.5-turbo-0125/nontoxic/gpt-3.5-turbo-0125_realtoxicityprompts_0shot_150_below_nontoxic_threshold_0_9_index.txt', 'r') as f:
    new_index = f.read().split()
    new_index = [int(i) for i in new_index]

print(set(new_index) - set(old_index)) # set() # 0.9 셋은 0.99 셋의 부분집합임
```

```python

## 0.99 셋 파일 중에, 0.9 셋에 들어가는 것들의 index를 구한다.
intersect_count = 0
intersect_index = []
for i, idx in enumerate(old_index):
    if idx in new_index:
        intersect_index.append(i)
        intersect_count+=1
print(intersect_count == len(new_index)) # True
```

```python
with open('new_module/llm_experiments/generate_with_llm/baselm_gens/gpt-3.5-turbo-0125/nontoxic/gpt-3.5-turbo-0125_realtoxicityprompts_0shot_150_below_nontoxic_threshold_0_95_to_0_9_index.txt', 'w') as f:
    f.write(' '.join([str(x) for x in intersect_index]))
    f.write('\n')
```
